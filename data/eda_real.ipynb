```python
# 주거와 일자리 데이터 분리
job_df = df[df['lclsfNm'].str.contains('일자리', na=False)]
house_df = df[df['lclsfNm'].str.contains('주거', na=False)]
```

```python
# 사용자는 자신의 조건에 맞는 정책을 더 클릭하고 선호할 것이다 -> 조건이 명확한 정책이 더 클릭될 가능성이 높다
```

```python
# 정책 수, 평균 조회수, 설명 길이 비교
print("일자리 정책 수:", len(job_df), "| 주거 정책 수:", len(house_df))
print("평균 조회수:", job_df['inqCnt'].mean(), "|", house_df['inqCnt'].mean())
print("설명 길이 평균:", job_df['plcyExplnCn'].astype(str).apply(len).mean(), "|", house_df['plcyExplnCn'].astype(str).apply(len).mean())
```

```python
# 소득 조건 유무에 따른 주거/일자리 조회수
house_df['has_income'] = house_df['earnMaxAmt'].notnull()
job_df['has_income'] = job_df['earnMaxAmt'].notnull()

print("주거 정책 - 소득 조건 유무에 따른 평균 조회수:")
print(house_df.groupby('has_income')['inqCnt'].mean())

print("일자리 정책 - 소득 조건 유무에 따른 평균 조회수:")
print(job_df.groupby('has_income')['inqCnt'].mean())
```

```python
# 일자리 주거 모두 소득 조건이 명시되어 있는게 조회수가 훨씬 높음
# 사람들은 조건을 진입장벽이라 생각하지 않는 것 같음
# 추천 시에 소득 조건 유무를 강조할 필요가 있음 (조건이 있으면 더 추천하기)
```

```python
# 최소나이/최대나이 조건 유무에 따른 주거/일자리 조회수
house_df['has_age'] = house_df['sprtTrgtMinAge'].notnull() | house_df['sprtTrgtMaxAge'].notnull()
job_df['has_age'] = job_df['sprtTrgtMinAge'].notnull() | job_df['sprtTrgtMaxAge'].notnull()

print("주거 정책 - 나이 조건 유무에 따른 평균 조회수:")
print(house_df.groupby('has_age')['inqCnt'].mean())

print("일자리 정책 - 나이 조건 유무에 따른 평균 조회수:")
print(job_df.groupby('has_age')['inqCnt'].mean())
```

```python
from collections import Counter
from itertools import chain
```

```python
# 일자리 키워드 Top 10
job_keywords = job_df['plcyKywdNm'].dropna().astype(str).str.split(',')
flat_job_kw = list(chain.from_iterable(job_keywords))
job_kw_counts = pd.Series(Counter(flat_job_kw)).sort_values(ascending=False)
top_job_kw = job_kw_counts.head(10)
print("일자리 키워드 Top 10 (빈도)")
print(top_job_kw)
```

```python
print("\n일자리 키워드별 평균 조회수 (내림차순 정렬):")

# 평균 조회수 계산해서 딕셔너리에 저장
job_kw_avg_views = {}
for kw in top_job_kw.index:
    mean_views = job_df[job_df['plcyKywdNm'].astype(str).str.contains(kw, na=False)]['inqCnt'].mean()
    job_kw_avg_views[kw] = mean_views

# 딕셔너리를 DataFrame으로 변환하고 정렬
job_kw_df = pd.DataFrame(list(job_kw_avg_views.items()), columns=['키워드', '평균 조회수'])
job_kw_df = job_kw_df.sort_values(by='평균 조회수', ascending=False)

# 출력
for _, row in job_kw_df.iterrows():
    print(f"{row['키워드']}: {row['평균 조회수']:.1f}")
```

```python
# 주거 키워드 Top 10
house_keywords = house_df['plcyKywdNm'].dropna().astype(str).str.split(',')
flat_house_kw = list(chain.from_iterable(house_keywords))
house_kw_counts = pd.Series(Counter(flat_house_kw)).sort_values(ascending=False)
top_house_kw = house_kw_counts.head(10)
print("\n주거 키워드 Top 10 (빈도)")
print(top_house_kw)
```

```python
print("\n주거 키워드별 평균 조회수 (내림차순 정렬):")

# 평균 조회수 계산해서 딕셔너리에 저장
house_kw_avg_views = {}
for kw in top_house_kw.index:
    mean_views = house_df[house_df['plcyKywdNm'].astype(str).str.contains(kw, na=False)]['inqCnt'].mean()
    house_kw_avg_views[kw] = mean_views

# 딕셔너리를 DataFrame으로 변환하고 정렬
house_kw_df = pd.DataFrame(list(house_kw_avg_views.items()), columns=['키워드', '평균 조회수'])
house_kw_df = house_kw_df.sort_values(by='평균 조회수', ascending=False)

# 출력
for _, row in house_kw_df.iterrows():
    print(f"{row['키워드']}: {row['평균 조회수']:.1f}")
```

```python
''' 
금전만 추천하는 건 안된다(언급을 잘하자로 가기- 피처 추가 가능)
있냐없냐 컬럼 제대로
만약 새로뽑았는데도 비율이 저러면 금전키워드+ 다른 임팩트 있는 피처 있을 확률
조건이 있으면 더 많이 주나? 
새로운 피처 발굴 필요
일자리만, 주거만, 따로  eda 진행
돈 말고 다른 중요한 피처
일자리 : 신입/경력자 그룹핑 -> 다시 중요도 파악
주거/지역별로 (서울/수도권/지방) -> 원하는 조건이 다를 수 있음
* 그룹핑을 어떻게 하냐에 따라 다르다
대분류 단위로 파일 쪼개기
마크다운으로 결론 정리
비율로 보기
'''
```

```python
from collections import Counter
from itertools import chain

# 키워드 평탄화
keywords_series = df['plcyKywdNm'].dropna().astype(str).str.split(',')
flat_keywords = list(chain.from_iterable(keywords_series))

# 상위 5개 키워드
top5_keywords = [kw for kw, _ in Counter(flat_keywords).most_common(5)]
```
