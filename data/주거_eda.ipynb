```python
# 대출 관련 정책 대상에 따라 조건 보기
def classify_target(text):
    if pd.isna(text):
        return '기타'
    elif '신혼' in text or '신혼부부' in text:
        return '신혼부부'
    elif '청년' in text:
        return '청년'
    else:
        return '기타'
```

```python
def classify_rent_support(row):
    text = (str(row['정책지원내용']) + ' ' + str(row['정책명'])).lower()

    if any(keyword in text for keyword in ['보증금', '월세', '대출', '이자', '지원금', '한도', '이차보전','임대료']):
        return '금전 지원'
    elif any(keyword in text for keyword in ['기숙사', '임대주택', '공공임대', '전세임대', '매입임대', '입주','주택 공급']):
        return '주택 제공'
    else:
        return '기타'
```

```python
df_rent['지원형태'] = df_rent.apply(classify_rent_support, axis=1)


# 결과 확인
df_rent['지원형태'].value_counts()

# 금전을 원하는지 주택을 원하는지 사용자에게 확인 필요
```

```python
import re

def extract_province(region_text):
    if pd.isna(region_text):
        return '기타'
    match = re.search(r'(서울|경기|인천|부산|대구|광주|대전|울산|세종|강원|충북|충남|전북|전남|경북|경남|제주|충청|경상|전국)', region_text)
    return match.group(0) if match else '기타'
```

```python
def map_to_area(province):
    if province == '서울':
        return '서울'
    elif province in ['경기', '인천']:
        return '수도권'
    elif province == '강원':
        return '강원'
    elif province in ['경북', '경남', '부산', '대구', '울산','경상']:
        return '경상도'
    elif province in ['충북', '충남', '대전', '세종','충청']:
        return '충청도'
    elif province == '제주':
        return '제주도'
    elif province == '전국':
        return '전국'
    else:
        return '기타'
```

```python
# 시·도 추출
df_housing['시도'] = df_housing['정책거주지역코드'].apply(extract_province)

# 도 단위 지역 그룹 생성
df_housing['도지역'] = df_housing['시도'].apply(map_to_area)
```
