```python
import pandas as pd
df = pd.read_csv(r"C:\dev\SKN10-FINAL-5Team\LLM\eval_qa_dataset.csv")
```

```python
df.info()
```

```python
# 평가지표 개수 확인
print("평가지표 개수:", df['평가지표'].nunique())
print("\n평가지표 목록:")
print(df['평가지표'].unique())
```

```python
# 평가지표별 질문 개수 계산
metric_counts = df['평가지표'].value_counts()

# 결과 출력
print("평가지표별 질문 개수:")
print(metric_counts)

# 시각화
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
metric_counts.plot(kind='bar')
plt.title('평가지표별 질문 개수')
plt.xlabel('평가지표')
plt.ylabel('질문 개수')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
